 DEFINING A BASIC FUNCTION

```python

def greet():
    print("Hello! Welcome to Python Functions.")
greet()
```

DEFINING A BASIC FUNCTION

```python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Alice")
greet_user("Bob")
```

 FUNCTION WITH DEFAULT PARAMETERS

```python
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()
greet_user("Ravi")
```

USING *args (VARIABLE NUMBER OF POSITIONAL ARGUMENTS)

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

 USING **kwargs (VARIABLE NUMBER OF KEYWORD ARGUMENTS)

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Alice", age=25, city="Delhi")
```

COMBINING *args AND **kwargs

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

 FUNCTION NESTING (CALL ONE FUNCTION INSIDE ANOTHER)


```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

```python
from functools import reduce

num = [1,2,3,4,5,6]
squares = list(map(lambda x: x**2,num))
even_squares = list(filter(lambda x: x % 2 == 0, squares))
print("squared list: ",squares)
print("Even squares: ",even_squares)
total = reduce(lambda x,y: x+y,squares)
print(total)
```

```python
d = {'a':2 , 'b': 3 , 'c': 4 }
for key,value in d.items():
  print(key,value)
squares = {key: value**2 for key, value in d.items()}
print(squares)
filter={key: value for key,value in squares.items()if value % 2 !=0}
print(filter)
total = reduce(lambda x,y:x+y,squares.values())
print(total)
```
