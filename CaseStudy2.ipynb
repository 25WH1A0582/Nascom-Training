Part F


```python
def F(x):
    return 40 + 5*x - 0.2*x**2
```

```python
def dF(x):
    return 5 - 0.4*x
```

```python
x = 5 / 0.4

print("x =", x)
```

```python
for x in range(1, 11):
    print("x =", x, "F(x) =", F(x), "F'(x) =", dF(x))
```

Part G

```python
def L(x):
    return 5*x - 0.2*x**2
```

```python
x = np.linspace(0, 5, 100)
```

```python
y = L(x)
```

```python
area = np.trapezoid(y, x)

print("Area:", area)
```

```python
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values

Y = df["Final_Score"].values
```

```python
X = np.column_stack((np.ones(len(X)), X))

print(X)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Coefficients:")
print(beta)
```
