```python
d={
    "h":[0,1,2,3,4,5],"c":[32,35,38,40,42,45],"v":[400,395,390,385,380,375],"a":[20,22,25,28,30,32],"e":[95,88,80,70,60,48]
}
```

```python
for j in range(6):
  print(d["h"][j],d["c"][j],d["v"][j],d["a"][j],d["e"][j])
```

```python
for j in range(6):
  if(d["c"][j]>40):
    print("High Temperature Alert")
  else:
    print("Temperature Normal")
```

```python
for j in range(6):
  if(d["e"][j]<50):
    print("Low Battery Alert")
  else:
    print("Battery Charge Normal")
```

```python
def battery_status(charge):
  if(charge>70):
    return "High"
  elif(charge>=50 and charge<=70):
    return "Medium"
  else:
    return "Low"

for j in range(6):
  a=battery_status(d["e"][j])
  print(a)
```
